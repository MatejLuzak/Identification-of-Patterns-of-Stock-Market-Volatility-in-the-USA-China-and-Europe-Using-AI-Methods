```python
import pandas as pd
```

```python
SP500 = pd.read_csv("sp500.csv")
```

```python
SP500 = SP500.iloc[2:].copy()
SP500.columns = ["Date", "Close", "High", "Low", "Open", "Volume"]
SP500.reset_index(drop=True, inplace=True)

SP500["Date"] = pd.to_datetime(SP500["Date"])
SP500 = SP500.astype({
    "Close": float,
    "High": float,
    "Low": float,
    "Open": float,
    "Volume": float
})
```
